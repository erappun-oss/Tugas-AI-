## 1. Studi Kasus: Metode Seismik Refraksi untuk Eksplorasi Air Tanah
Bertindaklah sebagai **[Role]** Ahli Geofisika Eksplorasi spesialis seismik dekat-permukaan. **[Task]** Analisislah data waktu tempuh (travel time) seismik refraksi untuk menentukan jumlah lapisan bawah permukaan dan kedalaman muka air tanah (akuifer). **[Context]** Survei ini dilakukan di daerah rawan kekeringan untuk mencari lokasi sumur bor yang optimal bagi warga desa. Berdasarkan **[Input]** grafik travel-time curve yang menunjukkan dua segmen kemiringan (segmen pertama $V_1$ = 400 m/s, segmen kedua $V_2$ = 1550 m/s dengan *intercept time* $t_i$ = 0.02 detik), kerjakan dengan **[Constraints]** menggunakan rumus metode *Intercept Time* sederhana dengan asumsi lapisan datar (horizontal), serta jelaskan perhitungan matematisnya tahap demi tahap. **[Output]** Hasilkan laporan singkat berisi estimasi kedalaman muka air tanah ($Z$) dan kesimpulan litologi batuan berdasarkan nilai kecepatannya. **[Quality Criteria]** Pastikan perhitungan matematis akurat dan interpretasi kecepatan 1550 m/s dikorelasikan secara logis dengan kecepatan tipikal zona jenuh air. Sebagai tambahan, **[Optional Verification]** berikan catatan disclaimer mengenai kemungkinan adanya blind-zone atau hidden layer yang tidak terdeteksi oleh metode survei ini.

## 2. Studi Kasus: Metode Gravitasi untuk Identifikasi Cekungan Sedimen
Kamu adalah **[Role]** Data Analyst Geofisika di sebuah perusahaan eksplorasi energi. **[Task]** Interpretasikan pola anomali Bouguer regional secara kualitatif untuk mendelineasi batas cekungan sedimen prospek hidrokarbon. **[Context]** Saat ini manajemen sedang mengevaluasi area *frontier* lepas pantai dan membutuhkan insight cepat sebelum memutuskan survei seismik 3D yang mahal. Diberikan **[Input]** deskripsi peta Anomali Bouguer Lengkap dengan zona anomali negatif oval (-45 mGal) yang dikelilingi tinggian anomali positif (+15 hingga +30 mGal). Dengan **[Constraints]** menghindari pemodelan numerik kuantitatif dan membatasi jawaban maksimal 3 paragraf (kurang dari 300 kata), berikan **[Output]** paragraf interpretasi struktur geologi bawah permukaan dan rekomendasi area fokus untuk survei seismik lanjutan. **[Quality Criteria]** Interpretasi harus secara jelas mengaitkan anomali rendah dengan akumulasi sedimen tebal berdensitas rendah, dan anomali tinggi dengan batuan dasar (basement) berdensitas tinggi. Terakhir, **[Optional Verification]** rekomendasikan prosedur verifikasi hasil menggunakan data sumur (*well log*) terdekat jika tersedia.

## 3. Studi Kasus: Geolistrik Resistivitas untuk Mitigasi Longsor
Sebagai **[Role]** Geoteknisi dan Ahli Geofisika Lingkungan, **[Task]** identifikasi kedalaman dan bentuk bidang gelincir (*slip surface*) dari profil penampang 2D geolistrik resistivitas. **[Context]** Pergerakan tanah perlahan baru saja terjadi di lereng perbukitan padat penduduk, dan BPBD meminta analisis cepat untuk menentukan batas aman evakuasi. Berdasarkan **[Input]** data profil 2D Inversi Resistivitas (ERT) yang menunjukkan lapisan atas (0-6 meter) bernilai 10-30 $\Omega m$ dan transisi tajam di bawah 6 meter menjadi 600-1200 $\Omega m$, buatlah analisis dengan **[Constraints]** menggunakan bahasa yang mudah dipahami oleh pejabat pemerintah non-teknis tanpa jargon ilmiah yang berlebihan. Hasilkan **[Output]** berupa memo teknis yang memuat letak bidang gelincir dan rekomendasi awal mitigasi bencana. **[Quality Criteria]** Memo harus menyoroti kontras resistivitas sebagai batas antara tanah penutup jenuh air/lapuk yang konduktif dengan batuan dasar masif yang resistif, disertai solusi mitigasi praktis. **[Optional Verification]** Beri penekanan bahwa hasil ini adalah estimasi geofisika yang tetap perlu divalidasi dengan pengeboran geoteknik di lapangan.

## 4. Studi Kasus: Metode Magnetik untuk Pemetaan Alterasi Panas Bumi
Bertindaklah sebagai **[Role]** Peneliti Geotermal yang ahli dalam metode potensial magnetik. **[Task]** Lakukan analisis fenomena anomali magnetik rendah (*low magnetic anomaly*) untuk mengidentifikasi zona alterasi hidrotermal. **[Context]** Kita sedang melakukan eksplorasi pendahuluan di kawasan gunung api purba untuk memetakan area potensi *cap rock* atau zona *upflow* panas bumi. Dengan **[Input]** peta kontur magnetik residual (RTP) yang menunjukkan zona anomali magnetik rendah signifikan di pusat area fumarol yang dikelilingi nilai tinggi dari batuan vulkanik sekitarnya, terapkan **[Constraints]** dengan memfokuskan penjelasan pada proses demagnetisasi batuan akibat fluida hidrotermal dan kaitannya dengan Titik Curie. Berikan **[Output]** berupa esai ilmiah singkat (400-500 kata) mengenai pembentukan anomali tersebut dan justifikasi area tersebut sebagai zona prospek. **[Quality Criteria]** Penjelasan harus mencakup konsep kimiawi tentang konversi mineral feromagnetik menjadi non-magnetik akibat reaksi hidrotermal suhu tinggi. **[Optional Verification]** Sertakan saran untuk melakukan *overlay* data magnetik dengan pemetaan geokimia untuk validasi silang.

## 5. Studi Kasus: Ground Penetrating Radar (GPR) untuk Kebocoran Pipa
Kamu ditugaskan sebagai **[Role]** Operator dan Interpreter Ground Penetrating Radar (GPR). **[Task]** Interpretasikan radargram GPR untuk mengidentifikasi posisi pipa air utama yang bocor di bawah jalan aspal. **[Context]** Dinas Pekerjaan Umum ingin mencari titik kebocoran di jalan yang sibuk tanpa melakukan penggalian acak yang akan merusak area terlalu luas. Berdasarkan **[Input]** pemindaian GPR (antena 500 MHz) yang memperlihatkan pola difraksi hiperbola kuat di kedalaman 1.2 meter dengan atenuasi sinyal (pelemahan drastis) di sekelilingnya hingga kedalaman 3 meter, patuhi **[Constraints]** untuk menghubungkan sifat fisik air (konduktivitas) dengan perilaku redaman sinyal gelombang elektromagnetik. Susun **[Output]** laporan inspeksi utilitas jalan yang menunjuk letak posisi pipa, estimasi kedalaman, dan alasan titik tersebut dicurigai bocor. **[Quality Criteria]** Analisis harus logis menunjukkan bahwa hiperbola mewakili pipa dan zona atenuasi di bawahnya membuktikan adanya genangan air yang menyerap energi radar. Sebagai langkah aman, **[Optional Verification]** instruksikan tim ekskavasi untuk melakukan *soft excavation* (penggalian manual yang hati-hati) pada kedalaman 1 meter agar pipa tidak pecah tersambar alat berat.
