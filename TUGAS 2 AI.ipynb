# HASIL ANALISIS STUDI KASUS 1
## Metode Seismik Refraksi untuk Eksplorasi Air Tanah

### 1. Pendahuluan

Metode seismik refraksi merupakan metode geofisika yang digunakan untuk mengidentifikasi struktur bawah permukaan berdasarkan perbedaan kecepatan rambat gelombang seismik pada setiap lapisan batuan. Kecepatan gelombang dipengaruhi oleh sifat fisik material, seperti elastisitas, kekompakan, porositas, dan tingkat kejenuhan air.

Dalam eksplorasi air tanah, metode ini dapat digunakan untuk memperkirakan kedalaman batas lapisan yang berpotensi menjadi zona jenuh air. Pada studi kasus ini, analisis dilakukan menggunakan metode *Intercept Time* berdasarkan kurva waktu tempuh (*travel-time curve*) yang menunjukkan dua segmen kemiringan.

### 2. Data dan Parameter

Berdasarkan hasil survei seismik refraksi, diperoleh data sebagai berikut:

| Parameter | Simbol | Nilai | Satuan |
|:---|:---:|:---:|:---:|
| Kecepatan lapisan pertama | $V_1$ | 400 | m/s |
| Kecepatan lapisan kedua | $V_2$ | 1550 | m/s |
| Waktu intersep | $t_i$ | 0,02 | s |
| Jumlah lapisan terinterpretasi | $n$ | 2 | Lapisan |
| Kedalaman batas antarlapisan | $Z$ | ? | m |

**Asumsi yang digunakan:**

1. Lapisan bawah permukaan bersifat horizontal.
2. Setiap lapisan dianggap homogen dengan kecepatan konstan.
3. Kecepatan gelombang meningkat terhadap kedalaman ($V_2 > V_1$).
4. Waktu intersep berasal dari gelombang refraksi kritis pada batas kedua lapisan.

### 3. Perhitungan Kedalaman Menggunakan Metode *Intercept Time*

#### 3.1 Persamaan Dasar

Untuk model dua lapisan horizontal, kedalaman batas antarlapisan dapat dihitung menggunakan persamaan:

$$
\boxed{Z=\frac{t_i V_1 V_2}{2\sqrt{V_2^2-V_1^2}}}
$$

**Keterangan:**

- $Z$ = Kedalaman batas antarlapisan (m)
- $t_i$ = Waktu intersep (s)
- $V_1$ = Kecepatan gelombang lapisan pertama (m/s)
- $V_2$ = Kecepatan gelombang lapisan kedua (m/s)

Persamaan tersebut diperoleh dari hubungan:

$$
t_i=\frac{2Z\cos i_c}{V_1}
$$

Dengan sudut kritis berdasarkan Hukum Snell:

$$
\sin i_c=\frac{V_1}{V_2}
$$

Sehingga:

$$
\cos i_c=\sqrt{1-\left(\frac{V_1}{V_2}\right)^2}
$$

#### 3.2 Tahapan Perhitungan

**Diketahui:**

$$
V_1=400\ \text{m/s}
$$

$$
V_2=1550\ \text{m/s}
$$

$$
t_i=0,02\ \text{s}
$$

**Ditanyakan:**

Kedalaman batas antarlapisan ($Z$).

**Penyelesaian:**

**Langkah 1.** Memasukkan nilai parameter ke dalam persamaan:

$$
Z=\frac{(0,02)(400)(1550)}
{2\sqrt{1550^2-400^2}}
$$

**Langkah 2.** Menghitung pembilang:

$$
(0,02)(400)(1550)=12400
$$

**Langkah 3.** Menghitung penyebut:

$$
1550^2=2402500
$$

$$
400^2=160000
$$

$$
\sqrt{2402500-160000}
=\sqrt{2242500}
$$

$$
\sqrt{2242500}\approx1497,50
$$

**Langkah 4.** Menghitung kedalaman:

$$
Z=\frac{12400}{2(1497,50)}
$$

$$
Z=\frac{12400}{2995,00}
$$

$$
\boxed{Z\approx4,14\ \text{m}}
$$

**Hasil:** Kedalaman batas antara lapisan pertama dan lapisan kedua diperkirakan sebesar **4,14 meter di bawah permukaan tanah**.

### 4. Interpretasi Geofisika dan Litologi

Berdasarkan kurva waktu tempuh yang menunjukkan dua segmen kemiringan, kondisi bawah permukaan dapat diinterpretasikan sebagai model dua lapisan.

| Lapisan | Kedalaman | Kecepatan Seismik | Interpretasi Litologi |
|:---:|:---:|:---:|:---|

#### 4.1 Lapisan Pertama

Lapisan pertama memiliki kecepatan gelombang seismik sebesar **400 m/s**.

Kecepatan yang relatif rendah menunjukkan kemungkinan keberadaan material dekat permukaan yang belum terkonsolidasi dengan baik, seperti tanah penutup, pasir lepas, atau material hasil pelapukan.

Material tersebut umumnya memiliki kekakuan relatif rendah sehingga gelombang seismik merambat lebih lambat.

#### 4.2 Lapisan Kedua

Lapisan kedua memiliki kecepatan gelombang seismik sebesar **1550 m/s**.

Nilai ini mendekati kecepatan gelombang kompresi dalam air, yaitu sekitar **1500 m/s** pada kondisi umum.

Kondisi tersebut mengindikasikan kemungkinan keberadaan material dengan tingkat kejenuhan air yang lebih tinggi, misalnya sedimen berpori yang terisi air tanah.

Namun, nilai kecepatan 1550 m/s tidak secara langsung membuktikan keberadaan akuifer karena kecepatan seismik juga dipengaruhi oleh kekompakan, jenis batuan, dan sifat elastis material.

Oleh karena itu, lapisan kedua diinterpretasikan sebagai **zona yang berpotensi jenuh air**, bukan sebagai akuifer yang telah terkonfirmasi.

### 5. Estimasi Kedalaman Muka Air Tanah

$$

Dengan demikian, **muka air tanah diperkirakan berada pada kedalaman sekitar 4,14 meter di bawah permukaan**, berdasarkan asumsi tersebut.

Akan tetapi, hasil ini masih merupakan interpretasi awal karena metode seismik refraksi mengidentifikasi perubahan kecepatan gelombang, bukan mengukur muka air tanah secara langsung.

### 6. Keterbatasan dan Verifikasi

Metode seismik refraksi memiliki keterbatasan berupa kemungkinan terjadinya *blind zone* atau *hidden layer*, yaitu kondisi ketika lapisan tertentu tidak dapat teridentifikasi dengan jelas melalui analisis waktu kedatangan pertama gelombang seismik.

Kondisi ini dapat terjadi apabila lapisan terlalu tipis, memiliki kecepatan lebih rendah dibandingkan lapisan di atasnya, atau tidak menghasilkan respons refraksi pertama yang dapat dibedakan.

Oleh karena itu, hasil interpretasi perlu diverifikasi menggunakan metode geolistrik resistivitas, pengukuran muka air sumur, atau pengeboran eksplorasi.

4. Hasil perhitungan menunjukkan bahwa batas kedua lapisan berada pada kedalaman sekitar **4,14 meter**.

5. Kedalaman tersebut dapat digunakan sebagai estimasi awal muka air tanah apabila peningkatan kecepatan terutama disebabkan oleh perubahan tingkat kejenuhan air.

6. Verifikasi menggunakan data geofisika tambahan dan pengeboran diperlukan untuk memastikan keberadaan akuifer serta menentukan lokasi sumur bor yang optimal.

**Kesimpulan utama:** Metode seismik refraksi menunjukkan adanya perubahan karakteristik material bawah permukaan pada kedalaman sekitar **4,14 meter**. ## Metode Gravitasi untuk Identifikasi Cekungan Sedimen

### Interpretasi Anomali Bouguer dan Rekomendasi Eksplorasi Hidrokarbon

**1. Interpretasi Zona Anomali Rendah**

Peta Anomali Bouguer Lengkap menunjukkan zona anomali negatif berbentuk oval dengan nilai mencapai **−45 mGal**. Anomali rendah ini mengindikasikan kemungkinan keberadaan cekungan sedimen yang memiliki densitas lebih rendah dibandingkan batuan di sekitarnya. Akumulasi sedimen yang relatif tebal dapat menghasilkan defisit massa bawah permukaan sehingga membentuk respons gravitasi negatif. Secara geologi, bagian tengah zona oval berpotensi merepresentasikan *depocenter*, yaitu bagian cekungan yang menjadi pusat pengendapan dan umumnya memiliki akumulasi sedimen lebih tebal. Namun, interpretasi tersebut masih bersifat awal karena anomali gravitasi juga dipengaruhi oleh variasi densitas batuan, kedalaman batuan dasar, dan struktur geologi regional.

**2. Interpretasi Zona Anomali Tinggi dan Struktur Geologi**

Zona anomali positif sebesar **+15 hingga +30 mGal** yang mengelilingi anomali negatif dapat diinterpretasikan sebagai kemungkinan tinggian batuan dasar (*basement high*) yang memiliki densitas relatif lebih besar dibandingkan sedimen pengisi cekungan. Kontras anomali tersebut menunjukkan adanya perbedaan distribusi massa bawah permukaan yang dapat berkaitan dengan perubahan kedalaman batuan dasar atau batas struktural cekungan. Daerah peralihan antara anomali rendah dan tinggi menjadi area penting untuk mengidentifikasi kemungkinan batas cekungan, sesar pembatas, serta konfigurasi struktur geologi yang berpotensi memengaruhi pembentukan perangkap hidrokarbon.

**3. Rekomendasi Survei Lanjutan dan Verifikasi**

Survei seismik lanjutan direkomendasikan pada **zona anomali negatif oval, khususnya bagian tengah cekungan dan daerah transisi menuju tinggian batuan dasar**. Bagian tengah diprioritaskan untuk mengevaluasi ketebalan sedimen, sedangkan zona transisi ditargetkan untuk mengidentifikasi sesar, geometri cekungan, dan kemungkinan perangkap struktural hidrokarbon. Sebelum pelaksanaan survei seismik 3D, disarankan melakukan interpretasi terpadu menggunakan data geologi regional dan seismik 2D jika tersedia. Interpretasi gravitasi juga perlu diverifikasi menggunakan data sumur terdekat, terutama *density log*, *sonic log*, dan informasi stratigrafi untuk mengevaluasi kontras densitas serta kedalaman batuan dasar. **Anomali gravitasi rendah menunjukkan potensi cekungan sedimen, tetapi tidak secara langsung membuktikan keberadaan akumulasi hidrokarbon.**

### 2. Interpretasi Kondisi Bawah Permukaan

| Parameter | Lapisan Pertama | Lapisan Kedua |

Air yang meresap ke dalam tanah dapat meningkatkan tekanan air dalam pori-pori. Dalam kondisi tertentu, peningkatan tekanan tersebut dapat mengurangi kekuatan tanah untuk menahan pergerakan, sehingga risiko longsor meningkat.

**Lapisan kedua (di bawah 6 meter)** memiliki resistivitas jauh lebih tinggi, yaitu 600–1200 Ωm. Nilai tersebut dapat mengindikasikan keberadaan batuan dasar yang lebih kompak, relatif masif, atau memiliki kandungan air yang lebih rendah.

Perbedaan karakteristik kedua lapisan menunjukkan adanya zona peralihan yang berpotensi menjadi bidang lemah tempat material tanah bagian atas bergerak mengikuti kemiringan lereng.

Berdasarkan perubahan resistivitas yang tajam pada kedalaman sekitar 6 meter, bidang gelincir potensial diinterpretasikan berada pada batas antara tanah penutup dan batuan dasar.

Secara konseptual, batas tersebut dapat menjadi permukaan tempat massa tanah bagian atas bergerak di atas lapisan batuan yang lebih kompak.

**Hasil estimasi:**

$$
\boxed{Z_{\text{bidang gelincir potensial}}\approx 6\ \text{meter}}
$$

Dengan:

- $Z$ = Kedalaman perkiraan bidang gelincir potensial.
- Lapisan atas = Material tanah lapuk atau berpotensi jenuh air.
- Lapisan bawah = Batuan dasar yang relatif kompak.

Apabila batas kedua lapisan mengikuti kemiringan lereng, bidang gelincir berpotensi memiliki bentuk yang relatif sejajar dengan permukaan batuan dasar. Namun, bentuk geometrinya, baik mendatar, melengkung, maupun tidak beraturan, **belum dapat dipastikan hanya berdasarkan rentang resistivitas dan kedalaman yang tersedia**.

Penentuan bentuk bidang gelincir secara lebih akurat memerlukan penampang resistivitas lengkap, informasi topografi, serta pemeriksaan kondisi tanah di lapangan.

**Catatan penting:** Kedalaman 6 meter merupakan estimasi batas perubahan sifat material bawah permukaan, bukan bukti langsung bahwa pergerakan tanah pasti terjadi tepat pada kedalaman tersebut.

## Metode Magnetik untuk Pemetaan Alterasi Panas Bumi

### Interpretasi Anomali Magnetik Rendah sebagai Indikator Zona Alterasi Hidrotermal pada Sistem Panas Bumi

#### 1. Pendahuluan

Metode magnetik merupakan salah satu metode geofisika yang digunakan dalam eksplorasi panas bumi untuk mengidentifikasi variasi sifat kemagnetan batuan bawah permukaan. Pada kawasan gunung api purba, aktivitas hidrotermal dapat menyebabkan perubahan mineralogi dan sifat magnetik batuan. Berdasarkan peta kontur anomali magnetik residual yang telah diproses menggunakan *Reduction to the Pole* (RTP), ditemukan zona anomali magnetik rendah yang signifikan di pusat area fumarol, dikelilingi anomali tinggi dari batuan vulkanik sekitarnya. Pola tersebut mengindikasikan kemungkinan adanya zona batuan yang mengalami penurunan magnetisasi akibat interaksi dengan fluida hidrotermal.

#### 2. Mekanisme Pembentukan Anomali Magnetik Rendah

Anomali magnetik rendah dapat terbentuk akibat proses demagnetisasi, yaitu berkurangnya magnetisasi batuan. Batuan vulkanik umumnya mengandung mineral magnetik seperti magnetit ($Fe_3O_4$) yang memiliki suseptibilitas magnetik relatif tinggi. Ketika fluida hidrotermal bersuhu tinggi bersirkulasi melalui rekahan batuan, interaksi antara fluida dan mineral dapat menyebabkan perubahan komposisi kimia serta mineralogi batuan. Dalam kondisi tertentu, magnetit dapat mengalami pelarutan atau transformasi menjadi mineral dengan sifat kemagnetan lebih lemah, seperti hematit ($Fe_2O_3$), pirit ($FeS_2$), atau mineral lempung tertentu. Proses tersebut dapat menurunkan kandungan mineral magnetik, suseptibilitas, dan magnetisasi batuan sehingga menghasilkan anomali magnetik rendah dibandingkan batuan vulkanik yang belum mengalami alterasi.

#### 3. Pengaruh Suhu dan Titik Curie

Selain perubahan mineralogi, temperatur tinggi dapat memengaruhi sifat kemagnetan batuan. Titik Curie merupakan temperatur ketika mineral feromagnetik atau ferimagnetik kehilangan keteraturan magnetisasi spontannya dan berubah menjadi bersifat paramagnetik. Magnetit memiliki Titik Curie sekitar **580°C**, sehingga pemanasan melampaui temperatur tersebut dapat menyebabkan hilangnya magnetisasi spontan selama mineral berada di atas Titik Curie. Namun, dalam sistem panas bumi, demagnetisasi tidak selalu memerlukan temperatur setinggi itu karena reaksi alterasi hidrotermal dapat mengurangi kandungan mineral magnetik pada temperatur yang lebih rendah. Oleh karena itu, anomali magnetik rendah dapat berkaitan dengan alterasi kimiawi, pengaruh termal, atau kombinasi keduanya.

#### 4. Justifikasi Zona Prospek Panas Bumi

Keberadaan anomali magnetik rendah yang berimpit dengan pusat fumarol memperkuat dugaan adanya aktivitas hidrotermal. Fumarol merupakan manifestasi permukaan yang menunjukkan keluarnya gas atau uap dari sistem panas bumi. Kombinasi manifestasi fumarol dan penurunan magnetisasi batuan dapat mengindikasikan jalur sirkulasi fluida panas atau zona alterasi hidrotermal yang berkembang. Area tersebut berpotensi berkaitan dengan zona *upflow*, yaitu jalur naiknya fluida panas, atau zona alterasi lempung yang dapat berperan sebagai *cap rock* apabila memiliki permeabilitas rendah. Namun, kedua kemungkinan tersebut belum dapat dibedakan hanya berdasarkan anomali magnetik. Dengan demikian, pusat anomali rendah layak diprioritaskan sebagai target eksplorasi lanjutan, tetapi belum dapat dipastikan sebagai lokasi reservoir panas bumi.

#### 5. Verifikasi dan Kesimpulan

Validasi interpretasi dapat dilakukan melalui *overlay* peta anomali magnetik residual RTP dengan data geokimia, khususnya distribusi gas fumarol, komposisi fluida, dan indikator alterasi hidrotermal. Integrasi dengan pemetaan geologi, identifikasi mineral alterasi, serta survei magnetotelurik (MT) juga diperlukan untuk mengevaluasi struktur dan distribusi resistivitas bawah permukaan. Secara keseluruhan, anomali magnetik rendah di pusat fumarol merupakan indikasi awal zona alterasi hidrotermal yang berpotensi berhubungan dengan sistem panas bumi. **Prospek tersebut memerlukan verifikasi multidisiplin sebelum penentuan target pengeboran eksplorasi.**

## Ground Penetrating Radar (GPR) untuk Identifikasi Kebocoran Pipa Air Bawah Tanah

### LAPORAN INSPEKSI UTILITAS JALAN

**Perihal:** Interpretasi Radargram GPR untuk Identifikasi Posisi Pipa dan Indikasi Kebocoran Air

| Pelaksana | Operator dan Interpreter Ground Penetrating Radar (GPR) |

| Kedalaman Indikasi Pipa | 1,2 meter |

| Tujuan Inspeksi | Identifikasi posisi pipa air utama dan indikasi kebocoran |

---

### 1. Pendahuluan

Ground Penetrating Radar (GPR) merupakan metode geofisika non-destruktif yang memanfaatkan gelombang elektromagnetik berfrekuensi tinggi untuk mendeteksi objek dan perubahan sifat fisik material bawah permukaan.

Metode ini banyak digunakan dalam inspeksi utilitas bawah tanah karena mampu mengidentifikasi keberadaan pipa, kabel, rongga, dan perubahan karakteristik material tanpa memerlukan penggalian langsung.

Pada studi kasus ini, survei GPR dilakukan menggunakan antena berfrekuensi **500 MHz** untuk mengidentifikasi posisi pipa air utama yang diduga mengalami kebocoran di bawah jalan aspal.

Hasil pemindaian menunjukkan adanya pola difraksi hiperbola kuat pada kedalaman sekitar **1,2 meter**, disertai zona pelemahan sinyal elektromagnetik (*attenuation*) di sekitarnya yang teramati hingga kedalaman sekitar **3 meter**.

Kombinasi kedua karakteristik tersebut menjadi dasar interpretasi posisi pipa dan kemungkinan keberadaan material bawah permukaan dengan kandungan air yang relatif tinggi.

### 2. Interpretasi Pola Difraksi Hiperbola

Pada radargram GPR, objek bawah permukaan yang berukuran relatif kecil terhadap lintasan pengukuran dapat menghasilkan pola refleksi berbentuk hiperbola.

Pola ini terbentuk karena jarak antara antena GPR dan objek berubah ketika antena bergerak melintasi objek tersebut.

Ketika antena berada tepat di atas objek, jarak tempuh gelombang elektromagnetik relatif paling pendek. Sebaliknya, ketika antena bergerak menjauhi objek, jarak tempuh gelombang meningkat sehingga waktu kedatangan pantulan menjadi lebih lama.

Perubahan waktu tempuh tersebut menghasilkan karakteristik refleksi berbentuk hiperbola pada radargram.

Dalam kasus ini, ditemukan **difraksi hiperbola kuat pada kedalaman sekitar 1,2 meter**.

Karakteristik tersebut konsisten dengan keberadaan objek utilitas bawah tanah, seperti pipa air.

Puncak hiperbola (*hyperbola apex*) dapat digunakan untuk memperkirakan posisi horizontal objek pada lintasan survei.

Dengan demikian, objek yang diduga sebagai pipa air utama diinterpretasikan berada pada:

$$
\boxed{Z_{\text{pipa}}\approx1,2\ \text{meter}}
$$

**Keterangan:**

- $Z_{\text{pipa}}$ = Estimasi kedalaman objek yang diduga pipa.
- Kedalaman 1,2 meter merupakan estimasi berdasarkan interpretasi radargram.
- Posisi horizontal pipa ditentukan dari lokasi puncak hiperbola pada lintasan pengukuran.

**Interpretasi utama:** Pola hiperbola kuat mengindikasikan keberadaan objek bawah permukaan yang konsisten dengan pipa air utama pada kedalaman sekitar 1,2 meter.

Namun, konfirmasi jenis objek tetap memerlukan pencocokan dengan peta jaringan utilitas atau pemeriksaan lapangan.

### 3. Analisis Atenuasi Gelombang Elektromagnetik

Selain pola hiperbola, radargram menunjukkan adanya zona pelemahan sinyal elektromagnetik yang signifikan di sekitar objek hingga kedalaman sekitar **3 meter**.

Fenomena ini dikenal sebagai *attenuation*, yaitu berkurangnya amplitudo atau energi gelombang elektromagnetik ketika merambat melalui suatu medium.

Besarnya atenuasi dipengaruhi oleh sifat kelistrikan material, terutama konduktivitas listrik, permitivitas dielektrik, dan frekuensi gelombang elektromagnetik.

Secara umum, medium dengan konduktivitas listrik lebih tinggi cenderung menyebabkan pelemahan gelombang elektromagnetik yang lebih besar.

Hubungan antara amplitudo gelombang dan atenuasi secara sederhana dapat dinyatakan sebagai:

$$
A(z)=A_0e^{-\alpha z}
$$

**Keterangan:**

- $A(z)$ = Amplitudo gelombang setelah merambat sejauh $z$.
- $A_0$ = Amplitudo awal gelombang.
- $\alpha$ = Koefisien atenuasi.
- $z$ = Jarak rambat gelombang dalam medium.

Persamaan tersebut menunjukkan bahwa amplitudo gelombang akan berkurang secara eksponensial seiring bertambahnya jarak rambat pada medium yang mengalami atenuasi.

Pada tanah, peningkatan kadar air dapat meningkatkan permitivitas dielektrik dan, bergantung pada kandungan ion terlarut serta karakteristik material, meningkatkan konduktivitas listrik.

Apabila konduktivitas meningkat, sebagian energi gelombang elektromagnetik mengalami disipasi sehingga sinyal radar yang diterima menjadi lebih lemah.

Dengan demikian, **zona atenuasi yang teramati di sekitar pipa dapat mengindikasikan keberadaan tanah dengan kandungan air relatif tinggi**.

### 4. Interpretasi Indikasi Kebocoran Pipa

Berdasarkan radargram, terdapat dua karakteristik utama yang mendukung dugaan kebocoran pipa.

**A. Keberadaan Objek yang Diduga Pipa**

Pola hiperbola kuat pada kedalaman sekitar 1,2 meter mengindikasikan adanya objek bawah permukaan yang konsisten dengan bentuk pipa.

Apabila objek tersebut sesuai dengan jalur pipa air utama berdasarkan peta utilitas, interpretasi keberadaan pipa menjadi lebih kuat.

**B. Zona Atenuasi di Sekitar Pipa**

Zona pelemahan sinyal yang teramati hingga kedalaman sekitar 3 meter dapat menunjukkan keberadaan material dengan sifat elektromagnetik berbeda dari lingkungan sekitarnya.

Salah satu kemungkinan penyebabnya adalah peningkatan kadar air tanah akibat rembesan dari pipa yang mengalami kebocoran.

Air yang meresap ke dalam tanah dapat mengubah sifat dielektrik dan konduktivitas material sehingga memengaruhi amplitudo gelombang radar.

Apabila air yang merembes mengandung ion terlarut, peningkatan konduktivitas dapat menyebabkan pelemahan sinyal elektromagnetik yang lebih signifikan.

Namun, zona atenuasi tidak secara langsung membuktikan adanya kebocoran karena respons serupa juga dapat disebabkan oleh tanah lempung, air tanah alami, atau material lain yang bersifat konduktif.

Oleh karena itu, hubungan antara keberadaan pipa dan zona atenuasi harus dipahami sebagai **indikasi kebocoran yang memerlukan verifikasi**, bukan bukti tunggal yang bersifat pasti.

### 5. Estimasi Posisi Pipa dan Zona yang Dicurigai Bocor

Berdasarkan interpretasi radargram GPR, diperoleh hasil sebagai berikut.

| Parameter | Hasil Interpretasi |
|:---|:---|
| Objek utama | Diduga pipa air bawah tanah |
| Kedalaman objek | Sekitar 1,2 meter |
| Indikator objek | Difraksi hiperbola kuat |
| Posisi horizontal | Berada pada lokasi puncak hiperbola di lintasan survei |
| Zona anomali tambahan | Atenuasi sinyal di sekitar objek |
| Kedalaman zona atenuasi | Hingga sekitar 3 meter |
| Dugaan penyebab atenuasi | Peningkatan kadar air atau keberadaan material konduktif |
| Interpretasi kebocoran | Indikasi awal yang memerlukan verifikasi |

Berdasarkan hasil tersebut, **area di sekitar puncak hiperbola pada kedalaman sekitar 1,2 meter direkomendasikan sebagai lokasi prioritas pemeriksaan kebocoran**.

Zona atenuasi hingga kedalaman sekitar 3 meter menunjukkan adanya perubahan respons elektromagnetik pada material di sekitar dan di bawah objek.

Namun, batas zona atenuasi tidak dapat langsung dianggap sebagai batas penyebaran air bocor karena dipengaruhi oleh sifat material dan kemampuan penetrasi sinyal GPR.

### 6. Rekomendasi Pemeriksaan dan Penanganan

Untuk memastikan keberadaan pipa serta kemungkinan kebocoran, direkomendasikan beberapa langkah berikut.

**A. Verifikasi Posisi Pipa**

1. Mencocokkan posisi hiperbola dengan peta jaringan utilitas bawah tanah.
2. Melakukan survei GPR tambahan menggunakan lintasan yang memotong perkiraan arah pipa.
3. Memastikan kedalaman objek melalui analisis kecepatan rambat gelombang elektromagnetik.
4. Menandai posisi horizontal objek berdasarkan puncak hiperbola pada radargram.

**B. Verifikasi Indikasi Kebocoran**

1. Melakukan pemeriksaan tekanan dan kehilangan air pada jaringan pipa.
2. Menggunakan metode pendeteksian kebocoran tambahan, seperti pemeriksaan akustik apabila sesuai dengan jenis pipa dan kondisi lapangan.
3. Membandingkan respons GPR pada zona yang dicurigai bocor dengan bagian pipa lain yang tidak menunjukkan anomali serupa.
4. Memeriksa kemungkinan sumber air lain, seperti air tanah alami, rembesan permukaan, atau saluran drainase.

**C. Rekomendasi Penggalian Aman**

Apabila hasil pemeriksaan tambahan mendukung dugaan kebocoran, penggalian terbatas dapat dilakukan pada area yang telah ditandai.

Penggalian harus dilakukan berdasarkan prosedur keselamatan utilitas dan pengaturan lalu lintas yang sesuai.

Mengingat kedalaman pipa diperkirakan sekitar **1,2 meter**, pekerjaan penggalian perlu dilakukan secara bertahap dengan metode *soft excavation* atau penggalian non-destruktif yang terkendali.

Pada kedalaman sekitar **1 meter**, tim harus meningkatkan kehati-hatian dan memastikan posisi aktual pipa sebelum melanjutkan penggalian.

Kedalaman tersebut merupakan titik kewaspadaan, bukan jaminan bahwa pipa berada tepat 20 cm di bawahnya.

Penggunaan alat berat di dekat posisi perkiraan pipa harus dibatasi sesuai prosedur keselamatan untuk menghindari kerusakan jaringan utilitas.

### 7. Keterbatasan Interpretasi

Interpretasi GPR memiliki beberapa keterbatasan yang perlu diperhatikan.

1. **Ketidakpastian kedalaman:** Estimasi kedalaman bergantung pada kecepatan gelombang elektromagnetik yang digunakan dalam konversi waktu tempuh menjadi kedalaman.

2. **Ambiguitas objek:** Pola hiperbola dapat dihasilkan oleh berbagai objek bawah permukaan, sehingga tidak selalu menunjukkan keberadaan pipa.

3. **Ambiguitas atenuasi:** Pelemahan sinyal dapat disebabkan oleh peningkatan kadar air, tanah lempung, material konduktif, atau kondisi geologi lainnya.

4. **Keterbatasan penetrasi:** Antena 500 MHz memberikan resolusi relatif tinggi, tetapi kemampuan penetrasinya dapat berkurang pada medium yang sangat konduktif.

5. **Keterbatasan identifikasi kebocoran:** Data GPR tidak dapat secara langsung menentukan laju kebocoran, kondisi fisik dinding pipa, maupun sumber pasti air yang menyebabkan perubahan respons elektromagnetik.

Dengan demikian, interpretasi harus dikombinasikan dengan data utilitas dan pemeriksaan lapangan sebelum dilakukan tindakan perbaikan.

### 8. Kesimpulan

Berdasarkan interpretasi radargram Ground Penetrating Radar (GPR) menggunakan antena 500 MHz, diperoleh kesimpulan sebagai berikut.

1. **Pola difraksi hiperbola kuat pada kedalaman sekitar 1,2 meter** mengindikasikan keberadaan objek bawah permukaan yang konsisten dengan pipa air utama.

2. Posisi horizontal objek dapat diperkirakan berdasarkan lokasi puncak hiperbola pada lintasan survei.

3. Zona atenuasi yang teramati hingga kedalaman sekitar **3 meter** mengindikasikan adanya perubahan sifat elektromagnetik material bawah permukaan.

4. Peningkatan kadar air tanah akibat kemungkinan kebocoran pipa dapat menyebabkan perubahan permitivitas dan konduktivitas listrik sehingga berkontribusi terhadap pelemahan sinyal GPR.

5. Kombinasi hiperbola dan zona atenuasi mendukung dugaan adanya kebocoran pipa, tetapi belum cukup untuk memastikan kebocoran tanpa pemeriksaan tambahan.

6. Pemeriksaan lanjutan direkomendasikan pada area di sekitar puncak hiperbola melalui verifikasi jaringan utilitas, pengujian kebocoran, dan penggalian terbatas yang aman apabila diperlukan.

> **REKOMENDASI UTAMA UNTUK DINAS PEKERJAAN UMUM**
>

Verifikasi menggunakan data jaringan utilitas dan metode pendeteksian kebocoran tambahan harus dilakukan sebelum penggalian. Apabila penggalian diperlukan, gunakan metode *soft excavation* yang terkendali dan tingkatkan kewaspadaan sebelum mencapai kedalaman perkiraan pipa untuk mencegah kerusakan utilitas.
